```python
#task3_1
def task3_1(quantity_of_data):
    if quantity_of_data[0:len(quantity_of_data) - 2].isdigit() == False:
        return "invalid data"
    elif quantity_of_data[len(quantity_of_data) - 1] != "B" :
        return "invalid data"
    elif quantity_of_data[len(quantity_of_data) - 2] == "K" :
        value = int(quantity_of_data[0:len(quantity_of_data) - 2]) * 1000
        return value
    elif quantity_of_data[len(quantity_of_data) - 2] == "M" :
        value = int(quantity_of_data[0:len(quantity_of_data) - 2]) * 1000000
        return value
    elif quantity_of_data[len(quantity_of_data) - 2] == "G" :
        value = int(quantity_of_data[0:len(quantity_of_data) - 2]) * 1000000000
        return value
    elif quantity_of_data[len(quantity_of_data) - 2] == "T" :
        value = int(quantity_of_data[0:len(quantity_of_data) - 2]) * 1000000000000
        return value
    else:
        return "invalid data"
    return


print(task3_1("8KB"))
print(task3_1("2MB"))
print(task3_1("90GB"))
print(task3_1("800TB"))
```

```python
#task3_2
def task3_2(quantity_of_data):
    if quantity_of_data[len(quantity_of_data)-1] == 'B':
        if quantity_of_data[len(quantity_of_data)-2] == 'i':
            return convert2(quantity_of_data)
        else:
            return task3_1(quantity_of_data)
    else:
        return 'invalid data'

def convert2(quantity_of_data):
    if quantity_of_data[0:len(quantity_of_data) - 3].isdigit() == False:
        return "invalid data"
    if quantity_of_data[len(quantity_of_data) - 3] == "K" :
        value = int(quantity_of_data[0:len(quantity_of_data) - 3]) * (2**10)
        return value
    elif quantity_of_data[len(quantity_of_data) - 3] == "M" :
        value = int(quantity_of_data[0:len(quantity_of_data) - 3]) *  (2**20)
        return value
    elif quantity_of_data[len(quantity_of_data) - 3] == "G" :
        value = int(quantity_of_data[0:len(quantity_of_data) - 3]) *  (2**30)
        return value
    elif quantity_of_data[len(quantity_of_data) - 3] == "T" :
        value = int(quantity_of_data[0:len(quantity_of_data) - 3]) *  (2**40)
        return value
    else:
        return "invalid data"


print(task3_2("8KB"))
print(task3_2("2MB"))
print(task3_2("90GB"))
print(task3_2("800TB"))
print(task3_2("8KiB"))
print(task3_2("2MiB"))
print(task3_2("90GiB"))
print(task3_2("800TiB"))
```
