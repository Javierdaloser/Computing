```python
#TASK 1.1
import sqlite3
connection = sqlite3.connect("TESTDATABASE.db")
connection.execute("DROP TABLE IF EXISTS Student")
connection.execute("DROP TABLE IF EXISTS Test")
connection.execute("DROP TABLE IF EXISTS Class")
connection.execute("DROP TABLE IF EXISTS Student_Test")

connection.execute( "CREATE TABLE Student" +
                   "(StudentID INTEGER PRIMARY KEY,\
                   GivenName TEXT,\
                   FamilyName TEXT,\
                   ClassName,\
                   FOREIGN KEY(ClassName) REFERENCES Class(ClassName)\
                   )")
connection.execute( "CREATE TABLE Test" +
                   "(TestID TEXT PRIMARY KEY,\
                   MaxMark INTEGER)")

connection.execute( "CREATE TABLE Class" +
                   "(ClassName TEXT PRIMARY KEY,\
                   YearGroup INTEGER\
                   )")
connection.execute( "CREATE TABLE Student_Test" +
                   "(StudentTestID INTEGER PRIMARY KEY,\
                   StudentID INTEGER,\
                   TestID TEXT,\
                   Mark INTEGER,\
                   DateTest TEXT,\
                   FOREIGN KEY(TestID) REFERENCES Test(TestID),\
                   FOREIGN KEY(StudentID) REFERENCES Student(StudentID)\
                   )")
connection.commit
connection.close
```
