```python
#task 2.2
def task2_2(list_of_integers):
    n = len(list_of_integers)
    if n == 1:
        return list_of_integers
    low = 0
    high = len(list_of_integers)
    mid = int(high/2)
    list_1 = list_of_integers[low:mid]
    list_2 = list_of_integers[mid:high]


    list_1 = task2_2(list_1)
    list_2 = task2_2(list_2)

    return mergesort(list_1,list_2)

def mergesort(a,b):
    i = 0
    j = 0
    k = 0
    c = [None] * (len(a) + len(b))
    while i < len(a) and j < len(b):
        if a[i] > b[j]:
            c[k] = b[j]
            k = k+1
            j=j+1
        elif a[i] < b[j]:
            c[k] = a[i]
            k = k+1
            i=i+1
        else:
            c[k] = a[i]
            c[k+1] = b[j]
            k = k+2
            i = i + 1
            j = j + 1
    while i < len(a):
        c[k] = a[i]
        k = k+1
        i=i+1
    while j < len(b):
        c[k] = b[j]
        k = k+1
        j=j+1
    return c


task2_2([56,25,4,98,0,18,4,5,7,0])
```

```python
#task2_3
def task2_3(filename_in, filename_out):
    with open(filename_in, 'r') as file:
        content = [int(x) for x in file.read().splitlines()]
        print(content)
        content = task2_2(content)
        print(content)
    with open(filename_out, 'w') as file:
        for num in content:
            file.write(str(num) + '\n')
    return
    
task2_3("randomnumbers_javier.txt", "sortednumbers_javier.txt")
```
